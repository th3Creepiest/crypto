```python
import math
from binance import Client
```

```python
client = Client()
```

```python
symbol_info = client.get_symbol_info("BTCUSDT")
symbol_info
```

```python
if symbol_info:
    lot_size = next((f for f in symbol_info["filters"] if f["filterType"] == "LOT_SIZE"), None)

lot_size
```

```python
if lot_size:
    step_size = float(lot_size["stepSize"])
    min_qty = float(lot_size["minQty"])

print("step size:", step_size)
print("min qty:", min_qty)
```

```python
precision = int(round(-math.log(step_size, 10), 0))
print("precision:", precision)
```

```python
quantity = 0.0012395

if quantity < min_qty:
    print("quantity is less than min qty")

print("quantity:", quantity)
```

```python
adjusted_quantity = math.floor(quantity / step_size) * step_size
adjusted_quantity = float("{:0.0{}f}".format(adjusted_quantity, precision))

if adjusted_quantity < min_qty:
    print("adjusted quantity is less than min qty")

print("adjusted quantity:", adjusted_quantity)
```
